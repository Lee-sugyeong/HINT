```python
df = pd.DataFrame({
    'temp': [22.0, None, 22.0, 31.5],
    'sensor': ['S1', 'S1', 'S1', 'S2']
})

print('결측치 수:\n', df.isna().sum())

df['temp'] = df['temp'].fillna(df['temp'].median())
df = df.drop_duplicates()

print(df)
```

```python
df = pd.DataFrame({
    'time': pd.to_datetime([
        '2026-08-01 09:00',
        '2026-08-01 09:05',
        '2026-08-01 10:00'
    ]),
    'temp': [22, 24, 26]
})

hourly = df.set_index('time').resample('1h')['temp'].mean()

print(hourly)
```

```python
df = pd.DataFrame({
    'temp': [22.0, None, 22.0, 31.5],
    'sensor': ['S1', 'S1', 'S1', 'S2']
})

print('결측치 수:\n', df.isna().sum())

df['temp'] = df['temp'].fillna(df['temp'].median())
df = df.drop_duplicates()

print(df)
```

```python
df = pd.DataFrame({'sensor_id': ['S1', 'S2', 'S3'],'temp': [23.0, 34.2, 41.0]})

def status(temp):
    if temp >= 40:
        return '위험'
    if temp >= 30:
        return '주의'
    return '정상'

df['status'] = df['temp'].apply(status)

print(df)
```

```python
df = pd.DataFrame({
    'time': pd.to_datetime([
        '2026-08-01 09:00',
        '2026-08-01 09:05',
        '2026-08-01 10:00'
    ]),
    'temp': [22, 24, 26]
})

hourly = df.set_index('time').resample('1h')['temp'].mean()

print(hourly)
```

```python
df = pd.DataFrame({'sensor_id': ['S1', 'S1', 'S2'],'temp': [22, 33, 35]})

summary = df.groupby('sensor_id',as_index=False)['temp'].agg(['mean', 'max']).reset_index()
alerts = df[df['temp'] >= 30]

summary.to_csv('sensor_summary.csv',index=False,encoding='utf-8-sig')
alerts.to_csv('sensor_alerts.csv',index=False,encoding='utf-8-sig')

print(summary)
```

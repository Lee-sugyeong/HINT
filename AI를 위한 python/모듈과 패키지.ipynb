```python
from pathlib import Path

project_dir = Path.cwd()/'project'
sensors_dir = project_dir/'sensors'

sensors_dir.mkdir(
    parents = True,
    exist_ok = True
)

print('프로젝트 :' , project_dir.resolve())
print('패키지 :', sensors_dir.resolve())
```

```python
converter_code = '''def c_to_f(celsius):
    """섭씨를 화씨로 변환합니다."""
    return celsius * 9 / 5 + 32
'''

(sensors_dir / 'converter.py').write_text(
    converter_code,
    encoding='utf-8'
)

print('converter.py 생성 완료')
```

```python
validator_code = '''def is_valid_temperature(value):
    """온도 값이 숫자인지 확인합니다."""
    return isinstance(value, (int, float))
'''

(sensors_dir / 'validator.py').write_text(
    validator_code,
    encoding='utf-8'
)

print('validator.py 생성 완료')
```

```python
init_code = '''# c_to_f 함수를 패키지 수준으로 공개합니다.
from .converter import c_to_f

# validator 모듈을 패키지 수준으로 공개합니다.
from . import validator
'''

(sensors_dir / '__init__.py').write_text(
    init_code,
    encoding='utf-8'
)

print('__init__.py 생성 완료')
```

```python
for path in project_dir.rglob('*'):
    print(path.relative_to(project_dir))
```

```python
import sys
import importlib

project_path = str(project_dir.resolve())

if project_path in sys.path:
    sys.path.remove(project_path)

sys.path.insert(0, project_path)

cached = [
    name for name in sys.modules
    if name == 'sensors' or name.startswith('sensors.')
]

for name in cached:
    del sys.modules[name]

importlib.invalidate_caches()

print('첫 검색 경로:', sys.path[0])
```

```python
from sensors.converter import c_to_f
from sensors import validator
import sensors

print('패키지 위치:', sensors.__file__)

temperature = 20

if validator.is_valid_temperature(temperature):
    fahrenheit = c_to_f(temperature)
    print(f'섭씨 {temperature}°C → 화씨 {fahrenheit}°F')
else:
    print('온도에 숫자를 입력하세요.')

# 정상 결과: 섭씨 20°C → 화씨 68.0°F
```

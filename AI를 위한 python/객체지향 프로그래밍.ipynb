```python
# 절차지향 방식
sensor = {'id': 'S1', 'temp': 24.5}

def show(sensor):
    print(f"{sensor['id']}:{sensor['temp']}")

# 객체지향 방식
class Sensor:
    def __init__(self, id, temp):
        self.id = id; self.temp = temp
    def show(self):
        print('{self.id}: {self.temp}')

Sensor('S1', 24.5).show()
```

```python
class Sensor:
    def __init__(self, sensor_id, value):
        self.sensor_id = sensor_id
        self.value = value
    def show(self):
        print(f'{self.sensor_id}: {self.value}')

s1 = Sensor('S-01', 24.5)
s2 = Sensor('S-02', 30.1)
s1.show()
s2.show()
```

```python
class Counter:
    def __init__(self, start=0):
        self.count = start

    def increase(self, step=1):
        self.count += step
        return self.count

counter = Counter()
print(counter.increase())
print(counter.increase(5))
```

```python
class Sensor:
    def __init__(self, value):
        self.value = value

    def add(self, amount):
        self.value += amount
        
s1 = Sensor(10)
s2 = Sensor(100)
s1.add(5)
print(s1.value, s2.value)
```

```python
class Sensor:
    count = 0

    def __init__(self, sensor_id):
        self.sensor_id = sensor_id
        Sensor.count += 1

s1 = Sensor('S1')
s2 = Sensor('S2')
print(Sensor.count)
print(s1.sensor_id, s2.sensor_id)
```

```python
class Sensor:
    def __init__(self, sensor_id):
        self.sensor_id = sensor_id

    def show(self):
        print(f"센서 ID: {self.sensor_id}")

class TemperatureSensor(Sensor):
    def to_fahrenheit(self, celsius):
        return celsius * 9 /5 + 32

ts = TemperatureSensor('T-01')
ts.show()
print(ts.to_fahrenheit(25))
```

```python
class Sensor:
    def __init__(self, sensor_id):
        self.sensor_id = sensor_id

class TemperatureSensor(Sensor):
    def __init__(self, sensor_id, unit='C'):
        super().__init__(sensor_id)
        self.unit = unit

ts = TemperatureSensor('T-01', unit='C')
print(ts.sensor_id, ts.unit)
```

```python
class Sensor:
    def read(self):
        return '측정값 없음'

class TemperatureSensor(Sensor):
    def read(self):
        return '24.5 C'

class HumiditySensor(Sensor):
    def read(self):
        return '60 %'

print(TemperatureSensor().read())
print(HumiditySensor().read()) 
```

```python
class Logger:
    def info(self):
        print('로그 기록')

class Sensor:
    def read(self):
        print('측정')

class SmartSensor(Logger, Sensor):
    pass

s = SmartSensor()
s.info()
s.read()
print(SmartSensor.__mro__)
```

```python
class Sensor:
    def __init__(self, sensor_id, value):
        self.sensor_id = sensor_id; self.value = value

    def __str__(self):
        return f'Sensor[{self.sensor_id}]={self.value}'

    def __eq__(self, other):
        return self.value == other.value

s1 = Sensor('S1', 24.5); s2 = Sensor('S2', 24.5)
print(s1)
print(s1==s2)
```

```python
import numpy as np
import pandas as pd

df = pd.DataFrame({'sensor' : ['S1', 'S2', 'S3'], 'temp' : [26.1, 21.5, 30.2], 'humidity' : [55, 61, 58]})
print(df.info())
```

```python
alerts = df[df['temp']>=30]
print(alerts)
```

```python
def status(temp):
    if temp >= 40:
        return '위험'
    if temp >= 30:
        return '주의'
    return '정상'

df['statue'] = df['temp'].apply(status)
print(df)
```

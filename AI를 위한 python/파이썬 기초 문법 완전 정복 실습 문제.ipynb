```python
temp = [10, 40, 20, 60, 100, 80, 34, 62]
for i in temp:
    if(i >= 100):
        level = "DANGER"
    elif(i >= 80)&(i < 100):
        level = "WARMING"
    else:
        level = "NORMAL"
    print(f"온도는 {i}도 {level}")
```
